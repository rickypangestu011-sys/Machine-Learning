```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from scipy.cluster.hierarchy import dendrogram, linkage
```

```python
df = pd.read_excel("nba_baru.xlsx")
df
```

```python
features1 = ["Height", "Weight"]

data = df [features1].copy()
data
```

```python
sns.scatterplot(data=df, x = "Height", y = "Weight")
plt.show()
```

```python
X=df['Height']
Y=df['Weight']

data = list(zip(X, Y))
print(data)
```

```python
#Single Linkage
linkage_data = linkage(data, method='single', metric='euclidean')
dendrogram(linkage_data)
plt.title("Single Linkage")

#batas threshold misalkan = 1
max_d=4.8
plt.axhline(y=max_d, c='k')
plt.show()
```

```python
#Complete Linkage
linkage_data1 = linkage(data, method='complete', metric='euclidean')
plt.title("Complete Linkage")
dendrogram(linkage_data1)
max_d=85
plt.axhline(y=max_d, c='k')
plt.show()
```

```python
#Average Linkage
linkage_data2 = linkage(data, method='average', metric='euclidean')
plt.title("Average Linkage")
dendrogram(linkage_data2)
max_d=30
plt.axhline(y=max_d, c='k')
plt.show()
```

```python
#Ward Linkage
linkage_data3 = linkage(data, method='ward', metric='euclidean')
plt.title("Ward Linkage")
dendrogram(linkage_data3)
max_d=260
plt.axhline(y=max_d, c='k')
plt.show()
```
