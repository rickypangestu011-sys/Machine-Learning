```python
import heapq
```

```python
def greedy(graph_data, h_value, source, destination):
    priority_queue = []
    heapq.heappush(priority_queue, (h_value[source], source, [source], 0))

    explored = set()

    while priority_queue:
        heuristic_value, current, route, total_cost = heapq.heappop(priority_queue)

        if current == destination:
            return route, total_cost

        if current in explored:
            continue

        explored.add(current)

        for next_node, distance in graph_data[current].items():
            if next_node not in explored:
                cost = total_cost + distance

                heapq.heappush(
                    priority_queue,
                    (
                        h_value[next_node],
                        next_node,
                        route + [next_node],
                        cost
                    )
                )

    return None, float('inf')
```

```python
# Data graf
graph_data = {
    'S': {'A': 10, 'B': 25, 'C': 30, 'D': 35, 'E': 10},
    'A': {'S': 10, 'B': 10, 'G': 90},
    'B': {'S': 25, 'A': 10, 'F': 5, 'K': 50},
    'C': {'S': 30, 'H': 40},
    'D': {'S': 35, 'E': 15, 'H': 25, 'L': 52},
    'E': {'S': 10, 'D': 15, 'J': 20},
    'F': {'B': 5, 'K': 40},
    'G': {'A': 90, 'K': 30, 'L': 40, 'M': 80},
    'H': {'C': 40, 'D': 25, 'L': 25},
    'J': {'E': 20, 'M': 40},
    'K': {'B': 50, 'F': 40, 'G': 30},
    'L': {'D': 52, 'H': 25, 'G': 40},
    'M': {'J': 40, 'G': 80}
}

# Nilai heuristik
h_value = {
    'S': 80,
    'A': 80,
    'B': 60,
    'C': 70,
    'D': 85,
    'E': 74,
    'F': 70,
    'G': 0,
    'H': 40,
    'J': 100,
    'K': 30,
    'L': 20,
    'M': 70
}
```

```python
# Menentukan titik awal dan tujuan
source = 'S'
destination = 'G'

route, total_cost = greedy(
    graph_data,
    h_value,
    source,
    destination
)

print("=== HASIL GREEDY SEARCH ===")

if route:
    print("Jalur Greedy :", " -> ".join(route))
    print("Total Cost   :", total_cost)
else:
    print("Jalur tidak ditemukan")
```
